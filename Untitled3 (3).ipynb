```python
"""
ADIM 1 - TAM KOD
CLT (Klasik Laminasyon Teorisi) - Q matrisi -> Q_bar -> ABD matrisi

Jupyter'da hucre hucre calistirmak icin asagidaki bloklari
ayri hucrelere kopyalayabilirsin (## HUCRE N ## yorumlariyla ayrildi).
"""

import numpy as np

# ----------------------------------------------------------------
# Malzeme sabitleri (tek yonlu karbon fiber/epoksi)
# Kaynak: Kaw, A.K., "Mechanics of Composite Materials", 2nd ed.,
#         CRC Press, 2006, Table 2.1 (Graphite/Epoxy, Vf=0.70)
#         (originally from Tsai & Hahn, "Introduction to Composite
#         Materials", Table 1.7, 7.1, 8.3)
# ----------------------------------------------------------------
E1 = 181e9      # Pa
E2 = 10.3e9     # Pa
G12 = 7.17e9    # Pa
nu12 = 0.28

Xt = 1500e6     # Pa
Xc = 1500e6     # Pa
Yt = 40e6       # Pa
Yc = 246e6      # Pa
S12 = 68e6      # Pa

## HUCRE 2 ##
# ----------------------------------------------------------------
# Lokal (1-2 eksen, fiber eksenine gore) Q matrisi
# ----------------------------------------------------------------
def compute_Q(E1, E2, G12, nu12):
    nu21 = nu12 * E2 / E1
    denom = 1 - nu12 * nu21

    Q11 = E1 / denom
    Q22 = E2 / denom
    Q12 = nu12 * E2 / denom
    Q66 = G12

    Q = np.array([
        [Q11, Q12, 0],
        [Q12, Q22, 0],
        [0,   0,   Q66]
    ])
    return Q
## HUCRE 3 ##
Q = compute_Q(E1, E2, G12, nu12)
print("Lokal Q matrisi (Pa):")
print(Q)

## HUCRE 4 ##
# ----------------------------------------------------------------
# Q matrisini laminat eksenine (global x-y) donduruyoruz -> Q_bar
# ----------------------------------------------------------------
def compute_Q_bar(Q, theta_deg):
    theta = np.radians(theta_deg)
    c = np.cos(theta)
    s = np.sin(theta)

    T = np.array([
        [c**2, s**2, 2*c*s],
        [s**2, c**2, -2*c*s],
        [-c*s, c*s, c**2 - s**2]
    ])

    Q_bar = T @ Q @ T.T
    return Q_bar


## HUCRE 5 ##
Q_bar_0 = compute_Q_bar(Q, 0)
print("0 derecede Q_bar:")
print(Q_bar_0)


## HUCRE 6 ##
Q_bar_90 = compute_Q_bar(Q, 90)
print("90 derecede Q_bar:")
print(Q_bar_90)


## HUCRE 7 ##
Q_bar_45 = compute_Q_bar(Q, 45)
print("45 derecede Q_bar:")
print(Q_bar_45)


## HUCRE 8 ##
# ----------------------------------------------------------------
# Laminat tanimi: her katmanin acisi ve kalinligi
# ----------------------------------------------------------------
layup = [
    {"angle": 0,  "thickness": 0.002},   # 2 mm
    {"angle": 90, "thickness": 0.002},
    {"angle": 0,  "thickness": 0.002},
]

total_thickness = sum(layer["thickness"] for layer in layup)
print("Toplam kalinlik:", total_thickness, "m")


## HUCRE 9 ##
# ----------------------------------------------------------------
# Her katmanin z_alt ve z_ust sinirlarini hesapla (orta yuzey z=0 referans)
# ----------------------------------------------------------------
z_start = -total_thickness / 2
z_positions = [z_start]
for layer in layup:
    z_start += layer["thickness"]
    z_positions.append(z_start)

for i, layer in enumerate(layup):
    layer["z_bottom"] = z_positions[i]
    layer["z_top"] = z_positions[i+1]
    print(f"Katman {i}: aci={layer['angle']}, z_alt={layer['z_bottom']:.4f}, z_ust={layer['z_top']:.4f}")


## HUCRE 10 ##
# ----------------------------------------------------------------
# ABD matrisi - kalinlik boyunca integrasyon
# A = membran (eksenel) rijitlik
# B = eksenel-egilme kuplaji
# D = egilme rijitligi
# ----------------------------------------------------------------
n = 3
A = np.zeros((n, n))
B = np.zeros((n, n))
D = np.zeros((n, n))

for layer in layup:
    Q = compute_Q(E1, E2, G12, nu12)
    Q_bar = compute_Q_bar(Q, layer["angle"])

    z_bot = layer["z_bottom"]
    z_top = layer["z_top"]

    A += Q_bar * (z_top - z_bot)
    B += Q_bar * (z_top**2 - z_bot**2) / 2
    D += Q_bar * (z_top**3 - z_bot**3) / 3

print("A matrisi (N/m):")
print(A)
print("\nB matrisi (N):")
print(B)
print("\nD matrisi (N.m):")
print(D)


## HUCRE 11 ##
# ----------------------------------------------------------------
# ORNEK: 100 kN eksenel yuk altinda panel ne kadar uzar?
# Panel genisligi = 1 m varsayimi -> N_x = 100 kN/m
# ----------------------------------------------------------------
N_applied = np.array([100e3, 0, 0])   # [Nx, Ny, Nxy], birim N/m
M_applied = np.array([0, 0, 0])       # moment yok

# 6x6 birlesik ABD sistemi (B sifir olsa bile genel dogru yontem)
ABD = np.block([[A, B], [B, D]])
NM = np.concatenate([N_applied, M_applied])

eps_kappa = np.linalg.solve(ABD, NM)
eps0 = eps_kappa[:3]     # orta yuzey birim sekildegistirmeleri
kappa = eps_kappa[3:]    # egrilikler

print("Orta yuzey sekildegistirmesi (eps0):", eps0)
print("Egrilik (kappa):", kappa)
```

```python
import numpy as np
import matplotlib.pyplot as plt

## HUCRE 1 ##
# ----------------------------------------------------------------
# Malzeme sabitleri (Adim 1 ile ayni)
# ----------------------------------------------------------------
E1 = 150e9
E2 = 10e9
G12 = 5e9
nu12 = 0.3

# Dayanim degerleri (Tsai-Wu icin gerekli) - tipik karbon/epoksi
Xt = 1500e6   # fiber yonunde cekme dayanimi (Pa)
Xc = 1200e6   # fiber yonunde basinc dayanimi
Yt = 50e6     # fibere dik cekme dayanimi
Yc = 200e6    # fibere dik basinc dayanimi
S12 = 70e6    # kayma dayanimi

```

```python
def compute_Q(E1, E2, G12, nu12):
    nu21 = nu12 * E2 / E1
    denom = 1 - nu12 * nu21
    Q11 = E1 / denom
    Q22 = E2 / denom
    Q12 = nu12 * E2 / denom
    Q66 = G12
    return np.array([[Q11, Q12, 0], [Q12, Q22, 0], [0, 0, Q66]])


def compute_Q_bar(Q, theta_deg):
    theta = np.radians(theta_deg)
    c, s = np.cos(theta), np.sin(theta)
    T = np.array([
        [c**2, s**2, 2*c*s],
        [s**2, c**2, -2*c*s],
        [-c*s, c*s, c**2 - s**2]
    ])
    return T @ Q @ T.T

```

```python
def build_layup(angles, thickness_each):
    layup = [{"angle": a, "thickness": thickness_each} for a in angles]
    total_t = sum(l["thickness"] for l in layup)
    z = -total_t / 2
    z_positions = [z]
    for l in layup:
        z += l["thickness"]
        z_positions.append(z)
    for i, l in enumerate(layup):
        l["z_bottom"] = z_positions[i]
        l["z_top"] = z_positions[i + 1]
        l["Q"] = compute_Q(E1, E2, G12, nu12)
        l["Q_bar"] = compute_Q_bar(l["Q"], l["angle"])
        l["failed"] = False          # hasar durumu (baslangicta hasarsiz)
        l["degrade_factor"] = 1.0    # rijitlik carpani (hasar sonrasi dusurulur)
    return layup
```

```python
# ABD matrisi hesabi - Adim 1 ile ayni mantik, ama artik degrade_factor
# katkisi da hesaba katiliyor (hasarli katmanin rijitligi dusuk sayilir)
# ----------------------------------------------------------------
def compute_ABD(layup):
    n = 3
    A = np.zeros((n, n))
    B = np.zeros((n, n))
    D = np.zeros((n, n))
    for l in layup:
        Q_eff = l["Q_bar"] * l["degrade_factor"]
        z_bot, z_top = l["z_bottom"], l["z_top"]
        A += Q_eff * (z_top - z_bot)
        B += Q_eff * (z_top**2 - z_bot**2) / 2
        D += Q_eff * (z_top**3 - z_bot**3) / 3
    return A, B, D

```

```python
def compute_ply_stresses(layup, eps0, kappa):
    results = []
    for l in layup:
        for z in [l["z_bottom"], l["z_top"]]:
            eps_xy = eps0 + z * kappa                        # global sekildegistirme
            sigma_xy = (l["Q_bar"] * l["degrade_factor"]) @ eps_xy   # global gerilme

            # global gerilmeyi katmanin lokal (1-2) eksenine dondur
            theta = np.radians(l["angle"])
            c, s = np.cos(theta), np.sin(theta)
            T_sigma = np.array([
                [c**2, s**2, 2*c*s],
                [s**2, c**2, -2*c*s],
                [-c*s, c*s, c**2 - s**2]
            ])
            sigma_12 = T_sigma @ sigma_xy

            results.append({
                "layer": l,
                "z": z,
                "sigma1": sigma_12[0],
                "sigma2": sigma_12[1],
                "tau12": sigma_12[2],
            })
    return results

```

```python
def tsai_wu_index(sigma1, sigma2, tau12):
    F1 = 1/Xt - 1/Xc
    F2 = 1/Yt - 1/Yc
    F11 = 1/(Xt*Xc)
    F22 = 1/(Yt*Yc)
    F66 = 1/(S12**2)
    F12 = -0.5 * np.sqrt(F11*F22)   # yaklasik etkilesim katsayisi

    index = (F1*sigma1 + F2*sigma2 + F11*sigma1**2 + F22*sigma2**2
             + F66*tau12**2 + 2*F12*sigma1*sigma2)
    return index

```

```python
angles = [0, 90, 0]
layup = build_layup(angles, thickness_each=0.002)

A, B, D = compute_ABD(layup)
N_applied = np.array([100e3, 0, 0])
M_applied = np.array([0, 0, 0])

ABD = np.block([[A, B], [B, D]])
NM = np.concatenate([N_applied, M_applied])
eps_kappa = np.linalg.solve(ABD, NM)
eps0 = eps_kappa[:3]
kappa = eps_kappa[3:]

print("--- 100 kN eksenel yuk altinda (hasarsiz durum) ---")
print("eps0:", eps0)
print("kappa:", kappa)

stresses = compute_ply_stresses(layup, eps0, kappa)
print("\nKatman gerilmeleri (lokal 1-2 eksininde):")
for r in stresses:
    idx = tsai_wu_index(r["sigma1"], r["sigma2"], r["tau12"])
    print(f"  aci={r['layer']['angle']:>3}, z={r['z']:.4f}: "
          f"sigma1={r['sigma1']/1e6:8.2f} MPa, sigma2={r['sigma2']/1e6:7.2f} MPa, "
          f"tau12={r['tau12']/1e6:7.2f} MPa, Tsai-Wu index={idx:.4f}")

```

```python
layup = build_layup(angles, thickness_each=0.002)   # temiz baslangic (hasarsiz)

N_steps = 500
N_max = 7000e3   # N/m
load_history = np.linspace(0, N_max, N_steps)

eps0_history = []
failure_log = []   # (yuk, katman_acisi, z) hasar anlari

for step, N_val in enumerate(load_history):
    A, B, D = compute_ABD(layup)
    ABD = np.block([[A, B], [B, D]])
    NM = np.array([N_val, 0, 0, 0, 0, 0])
    eps_kappa = np.linalg.solve(ABD, NM)
    eps0 = eps_kappa[:3]
    kappa = eps_kappa[3:]
    eps0_history.append(eps0[0])

    stresses = compute_ply_stresses(layup, eps0, kappa)
    for r in stresses:
        l = r["layer"]
        if not l["failed"]:
            idx = tsai_wu_index(r["sigma1"], r["sigma2"], r["tau12"])
            if idx >= 1.0:
                l["failed"] = True
                l["degrade_factor"] = 0.1   # rijitligi %90 dusur (ply-discount)
                failure_log.append((N_val, l["angle"], r["z"]))
                print(f"  >> HASAR: N={N_val/1e3:.1f} kN/m -> aci={l['angle']}, z={r['z']:.4f}")

print("\nToplam hasar olayi sayisi:", len(failure_log))
```

```python
eps0_history = np.array(eps0_history)
energy = np.trapezoid(load_history, eps0_history)

print(f"Toplam enerji absorpsiyonu: {energy:.2f} J/m (birim genislik basina)")

plt.figure(figsize=(7, 5))
plt.plot(eps0_history * 100, load_history / 1e3, 'b-', linewidth=2)   # eps0 -> yuzde
for (N_f, angle_f, z_f) in failure_log:
    idx = np.argmin(np.abs(load_history - N_f))
    plt.plot(eps0_history[idx] * 100, N_f / 1e3, 'rx', markersize=12, markeredgewidth=2.5)
    plt.annotate(f"{angle_f}° katman\nhasari\n(%{eps0_history[idx]*100:.2f})",
                 xy=(eps0_history[idx]*100, N_f/1e3),
                 xytext=(10, -25), textcoords='offset points', fontsize=8)
plt.xlabel("Orta yuzey birim sekildegistirmesi, eps0_x (%)")
plt.ylabel("Eksenel yuk, N_x (kN/m)")
plt.title(f"Laminat {angles} - Yuk-Sekildegistirme Egrisi\n(kirmizi X = katman hasari)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("yuk_sekildegistirme_adim2.png", dpi=150)
plt.show()
print("Grafik kaydedildi.")
```

```python
## YENİ BÖLÜM: RC FIBER-SECTION MODELİ (CLT ile karşılaştırma için)
# ----------------------------------------------------------------
# Beton ve donatı celigi icin basitlestirilmis gerilme-sekildegistirme modelleri
# ----------------------------------------------------------------

def concrete_stress(eps, fc=25e6, eps_c0=0.002, eps_cu=0.003):
    """
    Basitlestirilmis parabol-dikdortgen beton modeli (basinc pozitif kabul).
    fc: beton basinc dayanimi (Pa)
    eps_c0: pik gerilmedeki birim sekildegistirme
    eps_cu: nihai (ezilme) birim sekildegistirme
    """
    if eps <= 0:
        return 0.0  # cekmede beton catlar, katkisi yok (basitlestirme)
    if eps <= eps_c0:
        return fc * (2 * (eps / eps_c0) - (eps / eps_c0) ** 2)
    elif eps <= eps_cu:
        return fc  # plato varsayimi
    else:
        return 0.0  # ezilme sonrasi kapasite yok


def steel_stress(eps, fy=420e6, Es=200e9, eps_su=0.05):
    """
    Elastik - mukemmel plastik donati celigi modeli.
    fy: akma dayanimi, Es: elastisite modulu, eps_su: kopma birim sekildegistirmesi
    """
    eps_y = fy / Es
    if abs(eps) <= eps_y:
        return Es * eps
    elif abs(eps) <= eps_su:
        return np.sign(eps) * fy
    else:
        return 0.0  # kopma
```

```python
## Hucre: Kesit geometrisi ve fiber (lif) ayrimi
# ----------------------------------------------------------------
def build_fiber_section(width, height, n_fibers, rebar_layers):
    """
    width, height: kesit genisligi ve yuksekligi (m)
    n_fibers: beton icin lif sayisi (kalinlik boyunca)
    rebar_layers: [{"z": konum(m, orta eksenden), "area": donati alani(m2)}]
    """
    fiber_h = height / n_fibers
    z0 = -height / 2
    fibers = []
    
    for i in range(n_fibers):
        z_center = z0 + (i + 0.5) * fiber_h
        fibers.append({
            "z": z_center,
            "area": width * fiber_h,
            "material": "concrete",
            "failed": False,
        })
        
    for rb in rebar_layers:
        fibers.append({
            "z": rb["z"],
            "area": rb["area"],
            "material": "steel",
            "failed": False,
        })
        
    return fibers


# Fonksiyonu cagirma ve test etme
fibers = build_fiber_section(
    width=1.0, 
    height=0.20, 
    n_fibers=50,
    rebar_layers=[
        {"z": -0.09, "area": 4e-4},
        {"z": 0.09, "area": 4e-4}
    ]
)

eps0_max_rc = 0.006   # %0.6'ya kadar (beton ezilmesini de gormek icin)
N_steps = 50
eps0_history_rc = np.linspace(0, eps0_max_rc, N_steps)

print(fibers)
```

```python
## Hucre: Moment-egrilik dogrusal olmayan cozum (fiber-section)
# ----------------------------------------------------------------
def section_force(fibers, eps0, phi):
    """Verilen orta eksen sekildegistirmesi (eps0) ve egrilik (phi) icin
    kesit eksenel kuvveti (N) ve momentini (M) hesaplar.

    Hasar tespiti CLT tarafindaki (Hucre 8) l["failed"]=True mantigiyla
    birebir aynidir: bir fiber kritik sekildegistirmeyi astiginda acikca
    'failed' olarak bayraklanir ve rijitligi/katkisi dusurulur."""
    N, M = 0.0, 0.0
    for f in fibers:
        eps = eps0 + phi * f["z"]

        if f["material"] == "concrete":
            if eps > 0.003 and not f["failed"]:      # eps_cu = 0.003 (concrete_stress ile ayni)
                f["failed"] = True
            sigma = 0.0 if f["failed"] else concrete_stress(eps)
        else:
            sigma = steel_stress(eps)

        force = sigma * f["area"]
        N += force
        M += force * f["z"]
    return N, M


def solve_section(fibers, N_target, phi_guess=0.0, tol=1e-3, max_iter=50):
    """Verilen eksenel yuk N_target icin eps0'i Newton-Raphson ile bulur
    (basit sabit egrilik artimi altinda, egilme baskin durum icin)."""
    eps0 = 0.0
    for _ in range(max_iter):
        N_calc, _ = section_force(fibers, eps0, phi_guess)
        if abs(N_calc - N_target) < tol * abs(N_target + 1e-9):
            break
        d_eps = 1e-6
        N_plus, _ = section_force(fibers, eps0 + d_eps, phi_guess)
        dN_deps = (N_plus - N_calc) / d_eps
        if dN_deps == 0:
            break
        eps0 += (N_target - N_calc) / dN_deps
    return eps0
```

```python
## Hucre: RC fiber-section icin artimli yukleme (yer degistirme kontrollu) + CLT ile karsilastirma
# ----------------------------------------------------------------
N_history_rc = []

for eps0_val in eps0_history_rc:
    N_rc, _ = section_force(fibers, eps0_val, 0.0)
    N_history_rc.append(N_rc)

N_history_rc = np.array(N_history_rc)

# --- Karsilastirma grafigi: CLT vs RC fiber-section ---
plt.figure(figsize=(7, 5))
plt.plot(eps0_history * 100, load_history / 1e3, 'b-', linewidth=2, label="CLT (kompozit laminat)")
plt.plot(eps0_history_rc * 100, N_history_rc / 1e3, 'g--', linewidth=2, label="RC fiber-section")
plt.xlabel("Orta eksen birim sekildegistirmesi (%)")
plt.ylabel("Eksenel yuk, N (kN/m)")
plt.title("CLT vs RC Fiber-Section: Yuk-Sekildegistirme Karsilastirmasi")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("clt_vs_rc_karsilastirma.png", dpi=150)
plt.show()
```
